1. Find the Smallest and Largest Element in an Array

```python
def finsSmallest(lis):
    smallest = lis[0]
    for i in lis:
        if i < smallest:
            smallest = i

    return smallest

lis = [10, 20, 3, 4, 5]
li2 = [10,-1,-5,6,8]
finsSmallest(lis)
finsSmallest(li2)
```

#### Second Smallest

1. we get an array of values
2. need to find second smallest value, need to smallest value first and later find second smallest
3. take 2 variables in real time we don't know the values of array so always start smallelst with infinty value and second smalllest equal to  smallest
4. loop array of values and compare elemnest with smalllest value find smallest and assign smallest to second smallest value 
5. To find second smalllest value the condition is less than smallest and second smallest value 

```python
def findSecondSmallest(lis):
    smallest = float("inf")
    secondSmallest = smallest

    for i in lis:
        if i < smallest:
            secondSmallest = smallest
            smallest = i

    if (secondSmallest < i) and (i != secondSmallest):
        secondSmallest = i

    return secondSmallest


lis = [10, 2, 3, 5, -1, 2]
li1 = [10, 2, 3, 5, -1, 2,-5,-6]

print(findSecondSmallest(lis))
print(findSecondSmallest(li1))
```
